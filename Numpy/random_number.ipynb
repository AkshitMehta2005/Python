```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()

data = rng.normal(loc=70, scale=5, size=50)

plt.hist(data, bins=10)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()
```
