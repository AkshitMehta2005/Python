```python
import pandas as pd
df1 = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Aman"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
})

print(df1)
```

```python
df2 = pd.DataFrame({
    "Name": ["Neha", "Rohan", "Ankit"],
    "Age": [23, 21, 22],
    "Marks": [88, 76, 95]
})

print(df2)
```

```python
pd.concat([df1,df2])
```

```python
df1 = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Name": ["Rahul", "Priya", "Aman", "Neha"]
})

print(df1)
```

```python
df2 = pd.DataFrame({
    "Student_ID": [1, 2, 3, 5],
    "Marks": [85, 90, 78, 95]
})

print(df2)
```

```python
res = pd.merge(df1,df2,on="Student_ID")
res
```
